## Stationarity Test (Augmented Dickey-Fuller)
Most time series models require stationary data (constant mean and variance). Let's test the raw close price vs log returns.

```python
def print_adf_test(series, title):
    result = adfuller(series.dropna())
    print(f"--- ADF Test: {title} ---")
    print(f"ADF Statistic: {result[0]:.4f}")
    print(f"p-value: {result[1]:.4f}")
    print("Stationary" if result[1] < 0.05 else "Non-Stationary")
    print()

print_adf_test(df['close'], "Raw Close Price")

df['log_return'] = np.log(df['close'] / df['close'].shift(1))
print_adf_test(df['log_return'], "Log Returns")
```
