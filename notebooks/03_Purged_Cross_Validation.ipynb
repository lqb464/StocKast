# 03 — Purged Time Series Cross-Validation

**Goal**: Demonstrate how standard K-Fold leaks future information, and how Purged Time Series CV solves it.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.features.purged_cv import PurgedTimeSeriesSplit
from sklearn.model_selection import TimeSeriesSplit

sns.set_theme(style="darkgrid")
```

## Visualizing the Splits
Notice the gap (purge) between train and test sets to allow short-term correlations to decay.

```python
X = np.arange(100)
cv = PurgedTimeSeriesSplit(n_splits=4, purge_gap=5)

plt.figure(figsize=(10, 5))
for i, (train, test) in enumerate(cv.split(X)):
    plt.scatter(train, [i]*len(train), color='blue', marker='_', s=50, lw=5, label='Train' if i==0 else "")
    plt.scatter(test, [i]*len(test), color='red', marker='_', s=50, lw=5, label='Test' if i==0 else "")

plt.title('Purged Time Series Split (Gap=5)')
plt.ylabel('Split Fold')
plt.xlabel('Time Step')
plt.legend()
plt.yticks(range(4))
plt.show()
```
