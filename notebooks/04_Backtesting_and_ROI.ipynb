## Simulating Predictions
Let's create dummy predictions for the sake of the demonstration.

```python
df = pd.DataFrame({
    'date': pd.date_range('2023-01-01', periods=100),
    'log_return': np.random.normal(0.001, 0.02, 100),
    'pred_log_return': np.random.normal(0.001, 0.02, 100)
})

backtester = TradingBacktester(initial_capital=10000, transaction_cost=0.001)
metrics = backtester.run(df, pred_col="pred_log_return")
print(metrics)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['portfolio_value'], label="Strategy Portfolio Value")
plt.axhline(10000, color='r', linestyle='--', label="Initial Capital")
plt.title('Backtest Results: Portfolio Value over Time')
plt.legend()
plt.show()
```
