# 02 — Advanced Feature Engineering

**Goal**: Extract powerful technical indicators like MACD, RSI, ATR, and Bollinger Bands.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loader import fetch_stock_data
from src.features.technical import add_technical_indicators, generate_targets

sns.set_theme(style="darkgrid")
```

```python
df = fetch_stock_data("AAPL", period="2y")
df = add_technical_indicators(df)
df = generate_targets(df, horizon=1)
df = df.dropna()
df.tail()
```

```python
fig, ax = plt.subplots(3, 1, figsize=(14, 12), sharex=True)

ax[0].plot(df['date'], df['close'], label='Close')
ax[0].plot(df['date'], df['bb_upper'], 'r--', alpha=0.5)
ax[0].plot(df['date'], df['bb_lower'], 'r--', alpha=0.5)
ax[0].set_title('Price & Bollinger Bands')

ax[1].plot(df['date'], df['macd'], label='MACD')
ax[1].plot(df['date'], df['macd_signal'], label='Signal')
ax[1].set_title('MACD')
ax[1].legend()

ax[2].plot(df['date'], df['rsi_14'], label='RSI')
ax[2].axhline(70, color='r', linestyle='--')
ax[2].axhline(30, color='g', linestyle='--')
ax[2].set_title('RSI (14)')

plt.tight_layout()
plt.show()
```
